```python
import pandas as pd

# Load the dataset
df = pd.read_csv('Finance_Accounting_Dataset.csv')

# Filter out rows where Assets, Liabilities, Market_Cap, Share_Price and Expenses have negative values
df = df[(df['Assets'] >= 0) & (df['Liabilities'] >= 0) &
        (df['Market_Cap'] >= 0) & (df['Share_Price'] >= 0) & (df['Expenses'] >= 0)]

# Save the result as a CSV file
csv_path = 'Finance_Accounting_Dataset_cleaned.csv'
df.to_csv(csv_path, index=False)
```
